INTRODUCTION

1.1 Background of the Study

The rapid advancement of digital technologies has transformed the landscape of education, making online learning one of the most significant innovations in contemporary teaching and learning. Educational institutions, universities, corporate training organizations, and Massive Open Online Course (MOOC) providers increasingly rely on online learning platforms to deliver flexible, accessible, and cost-effective educational opportunities. These platforms enable learners to access instructional materials regardless of geographical location and time constraints, thereby promoting lifelong learning and widening educational access.

The global adoption of online learning accelerated significantly during the COVID-19 pandemic, when schools and universities transitioned from traditional face-to-face instruction to fully online or hybrid learning environments. Even after the pandemic, many institutions have continued integrating online learning into their educational systems because of its flexibility, scalability, and ability to support personalized learning experiences. Consequently, online learning has evolved from an alternative mode of instruction into an integral component of modern education.

Despite these advantages, online learning providers continue to face a persistent challenge: low course completion rates. Studies consistently report that many learners enroll in online courses but fail to complete them due to various academic, personal, technological, and motivational factors. Common reasons include inadequate time management, declining motivation, limited engagement with course materials, insufficient interaction with instructors and peers, poor internet connectivity, competing professional or family responsibilities, and a lack of prior online learning experience. These challenges not only affect learner achievement but also reduce the effectiveness and credibility of online education programs.

Educational institutions increasingly recognize that improving learner retention requires more than simply providing access to digital content. Instead, institutions must proactively identify learners who are at risk of dropping out and implement timely interventions before disengagement occurs. Traditional methods of monitoring student progress often rely on manual observation or periodic academic evaluations, which may fail to detect early warning signs. As a result, there is growing interest in applying educational data analytics and artificial intelligence to support evidence-based decision-making in online education.

Machine learning has emerged as a powerful tool for extracting meaningful insights from large educational datasets. By analyzing historical learner data, machine learning algorithms can identify complex relationships among demographic characteristics, academic performance, learning behaviors, and engagement indicators that influence student success. Unlike traditional statistical methods that often assume linear relationships, machine learning techniques can model complex and nonlinear interactions, making them particularly suitable for predicting educational outcomes.

Predictive analytics in education has gained considerable attention over the past decade. Previous studies have demonstrated that variables such as quiz performance, assignment completion, participation in online discussions, video viewing behavior, study habits, motivation, and previous academic achievement significantly influence course completion. Machine learning models can leverage these variables to generate predictions regarding learner success, enabling educators to implement targeted interventions such as personalized tutoring, academic advising, motivational support, or adaptive learning strategies.

However, predictive performance alone is insufficient for educational decision-making. Educational institutions require models that are not only accurate but also transparent, interpretable, and fair. Recent advances in Explainable Artificial Intelligence (XAI), including SHapley Additive Explanations (SHAP) and Local Interpretable Model-Agnostic Explanations (LIME), have made it possible to explain why machine learning models produce specific predictions. These techniques improve stakeholder trust by identifying the factors that most strongly influence prediction outcomes.

Similarly, the increasing adoption of AI in education has raised concerns regarding fairness and bias. Machine learning models trained on historical educational data may unintentionally disadvantage certain demographic groups if underlying biases exist within the dataset. Consequently, responsible AI practices require fairness auditing using metrics such as demographic parity, equalized odds, and disparate impact to ensure that predictive models support equitable educational opportunities.

This study addresses these challenges by developing and evaluating machine learning classification models to predict students' course completion status using the Online Learning Dataset 2026 (https://www.kaggle.com/datasets/patelris/smart-learning-recommendation-dataset/data). The study compares multiple classification algorithms, including Logistic Regression, Decision Tree, Random Forest, Support Vector Machine, and Extreme Gradient Boosting (XGBoost), to determine the most effective predictive model. Furthermore, the selected model will be interpreted using Explainable AI techniques and evaluated through fairness analysis to ensure transparency and ethical application.

The findings of this study are expected to contribute to the growing field of educational data mining by demonstrating how machine learning can support early identification of at-risk learners, improve student retention strategies, and enhance decision-making for online learning providers. Ultimately, the proposed predictive framework seeks to promote learner success while encouraging the responsible adoption of artificial intelligence in education.

# 1.2 Problem Understanding and Framing

## 1.2.1 Business Problem

Online learning has become an integral component of modern education because it provides learners with flexible, accessible, and personalized learning opportunities. However, despite its widespread adoption, learner retention remains a significant challenge for educational institutions and online learning providers. Many students discontinue their courses before completion due to factors such as limited engagement, inadequate motivation, poor time management, technological barriers, and competing personal or professional commitments. High dropout rates reduce course completion rates, increase the demand for learner support services, and negatively affect institutional effectiveness and learner satisfaction.

Educational institutions require proactive approaches that can identify learners who are at risk of dropping out before disengagement occurs. Traditional monitoring methods, which rely primarily on periodic assessments and manual observation, often fail to detect early warning signs of learner withdrawal. Consequently, there is a need for a predictive decision-support system that enables educators to identify at-risk learners and implement timely interventions such as academic advising, mentoring, and personalized learning support. The central business problem addressed in this study is therefore the need to improve learner retention by accurately predicting whether a learner will successfully complete an online course or drop out.



## 1.2.2 Data Science Problem

From a data science perspective, the problem involves developing a predictive machine learning model capable of classifying learners according to their final course outcome using historical learner data. The dataset contains demographic information, academic performance indicators, engagement metrics, technological characteristics, and personal learning attributes that collectively influence learner persistence in online education.

Although the original dataset includes three completion status categories (**Completed**, **Dropped**, and **In Progress**), this study focuses exclusively on learners whose final outcomes are known. Records labeled **In Progress** were excluded because they do not represent completed learning outcomes and would introduce uncertainty into the target variable. Consequently, the machine learning models are trained using only observations classified as **Completed** or **Dropped**, ensuring that each training instance has a definitive outcome.

The predictive model is intended not only to achieve high classification performance but also to support evidence-based educational decision-making. Accordingly, Explainable Artificial Intelligence (XAI) techniques will be employed to improve model transparency and enable educators to understand the factors influencing individual predictions.



## 1.2.3 Machine Learning Task

This study is formulated as a **supervised binary classification** problem.

The target variable is **completion_status**, representing the learner's final course outcome. After excluding learners with an **In Progress** status, the target variable consists of two mutually exclusive classes:

* **Completed**
* **Dropped**

Binary classification is the most appropriate machine learning approach because the objective is to predict one of two predefined categorical outcomes using historical labeled data. Regression techniques are unsuitable because the target variable is categorical rather than continuous, while clustering methods are intended to discover naturally occurring groups in unlabeled data instead of predicting known outcomes.



## 1.2.4 Success Metrics

The predictive performance of the machine learning models will be evaluated using several complementary classification metrics to ensure a comprehensive assessment of model effectiveness.

**Accuracy** will measure the overall proportion of correctly classified learners. **Precision** will evaluate the proportion of learners predicted to have completed the course who actually completed it, while **Recall** will measure the model's ability to correctly identify learners who successfully completed their courses. **F1-Score**, which represents the harmonic mean of Precision and Recall, will serve as the primary technical evaluation metric because it provides a balanced assessment of model performance, particularly when class distributions are unequal. In addition, the **Area Under the Receiver Operating Characteristic Curve (ROC-AUC)** will be used to evaluate the model's ability to distinguish between learners who complete their courses and those who drop out across different classification thresholds. Confusion matrices and class-specific performance reports will also be analyzed to assess model performance for both learner outcomes.



## 1.2.5 Business Key Performance Indicators

The practical value of the predictive model will be assessed using business-oriented key performance indicators that reflect improvements in educational effectiveness and learner success.

The primary business objective is to **increase course completion rates** by enabling educational institutions to identify learners who may require additional academic support before they disengage from their courses. Successful implementation of the predictive model is also expected to contribute to **reduced learner dropout rates**, **improved learner retention**, and **more efficient allocation of academic support resources** by allowing educators to prioritize interventions for learners who exhibit characteristics associated with unsuccessful course completion. Ultimately, these improvements are expected to enhance learner satisfaction, strengthen institutional performance, and promote data-driven decision-making within online learning environments.



## 1.2.6 Capstone Deliverable

The primary deliverable for this phase is a clearly defined machine learning problem that aligns educational objectives with predictive analytics. Specifically, this step establishes the business problem, formulates the corresponding data science problem, identifies the machine learning task as **supervised binary classification**, defines the target variable (**completion_status**), and specifies both the technical success metrics (Accuracy, Precision, Recall, F1-Score, and ROC-AUC) and the business key performance indicators (course completion rate, dropout rate, learner retention, resource optimization, and learner satisfaction).
```python
import numpy as np

def summarize_features(df):
    feature_summary = []
    for col in df.columns:
        col_info = {
            'Feature Name': col,
            'Data Type': str(df[col].dtype),
            'Missing Values': df[col].isnull().sum()
        }

        if pd.api.types.is_numeric_dtype(df[col]):
            col_info['Category'] = 'Numerical'
            if df[col].nunique() < 20 and all(df[col].dropna().apply(lambda x: x == int(x))):
                col_info['Sub-Category'] = 'Discrete'
                col_info['Description'] = 'Counts or codes with a limited number of unique integer values.'
                col_info['Characteristics'] = f"Unique Values: {df[col].unique().tolist() if df[col].nunique() < 20 else 'Too many to list'}"
            else:
                col_info['Sub-Category'] = 'Continuous'
                col_info['Description'] = 'Measurements or quantities that can take any value within a range.'
                desc_stats = df[col].describe().loc[['mean', 'std', 'min', 'max']].to_dict()
                col_info['Characteristics'] = f"Mean: {desc_stats['mean']:.2f}, Std: {desc_stats['std']:.2f}, Min: {desc_stats['min']:.2f}, Max: {desc_stats['max']:.2f}"
        else:
            col_info['Category'] = 'Categorical'
            if df[col].nunique() / len(df) > 0.5: # Arbitrary threshold for high cardinality
                col_info['Sub-Category'] = 'Nominal (High Cardinality)'
                col_info['Description'] = 'Categories without inherent order, many unique values. Often identifiers.'
            elif df[col].nunique() <= 10:
                col_info['Sub-Category'] = 'Nominal/Ordinal'
                col_info['Description'] = 'Categories without/with inherent order, few unique values.'
            else:
                col_info['Sub-Category'] = 'Nominal'
                col_info['Description'] = 'Categories without inherent order.'
            col_info['Characteristics'] = f"Unique Values: {df[col].unique().tolist()}"

        feature_summary.append(col_info)
    return pd.DataFrame(feature_summary)

summary_df = summarize_features(df)

print("### Feature Type Summary\n")
print(summary_df.to_markdown(index=False))

# Re-confirming missing values across all columns
missing_values = df.isnull().sum()
if missing_values.sum() == 0:
    print("\n\n### Missing Values Re-confirmation\n")
    print("No missing values found across any columns. Data completeness is confirmed.")
else:
    print("\n\n### Missing Values Re-confirmation\n")
    print("Missing values detected in the following columns:\n")
    print(missing_values[missing_values > 0].to_markdown())
```

## 3.1.6 Detect Outliers using IQR



**Reasoning**:
Implementing the IQR method to detect outliers in each numerical continuous column and print the results.



```python
print("### Outlier Detection using IQR Method\n")
outlier_results = {}

for col in numerical_continuous_cols:
    Q1 = df_processed[col].quantile(0.25)
    Q3 = df_processed[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify outliers
    outliers = df_processed[(df_processed[col] < lower_bound) | (df_processed[col] > upper_bound)]
    num_outliers = outliers.shape[0]
    percentage_outliers = (num_outliers / len(df_processed)) * 100

    outlier_results[col] = {
        'num_outliers': num_outliers,
        'percentage_outliers': percentage_outliers
    }

    print(f"Column: {col}")
    print(f"  Q1: {Q1:.2f}")
    print(f"  Q3: {Q3:.2f}")
    print(f"  IQR: {IQR:.2f}")
    print(f"  Lower Bound: {lower_bound:.2f}")
    print(f"  Upper Bound: {upper_bound:.2f}")
    print(f"  Number of Outliers: {num_outliers}")
    print(f"  Percentage of Outliers: {percentage_outliers:.2f}%\n")

print("Outlier detection complete for all numerical continuous columns.")
```

### Summary of Outlier Detection Results

The summary of the outlier detection results, including identification of columns with outliers and their potential implications, has been thoroughly addressed in the previous subtask's summary. To reiterate the key findings:

## 3.5 Dimensionality Reduction (PCA)

Principal Component Analysis (PCA) is a powerful technique for dimensionality reduction. It transforms a large set of variables into a smaller one that still contains most of the information in the large set. This can help to simplify models, reduce noise, and improve computational efficiency. Here, we'll apply PCA to the complete preprocessed feature set (`X`) to explore its structure.

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import numpy as np

# Initialize PCA
pca = PCA()

# Fit PCA to the full preprocessed feature set (X)
pca.fit(X)

# Plot the cumulative explained variance ratio
plt.figure(figsize=(10, 6))
plt.plot(np.cumsum(pca.explained_variance_ratio_))
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Cumulative Explained Variance by Number of Principal Components')
plt.grid(True)
plt.show()

print("Explained variance plot generated.")
```

### Interpreting the Explained Variance Plot

The plot above shows the cumulative explained variance as a function of the number of principal components. This helps us determine how many components are needed to capture a certain percentage of the total variance in the dataset. For instance, if the curve flattens significantly after a certain number of components, it suggests that adding more components beyond that point provides diminishing returns in terms of explaining additional variance.

```python
import seaborn as sns

# Reduce dimensions to the first two principal components for visualization
pca_2d = PCA(n_components=2)
principal_components = pca_2d.fit_transform(X)

# Create a DataFrame for the principal components and the target variable
pca_df = pd.DataFrame(data=principal_components, columns=['Principal Component 1', 'Principal Component 2'])
pca_df['completion_status'] = y.values

# Visualize the first two principal components
plt.figure(figsize=(10, 8))
sns.scatterplot(
    x='Principal Component 1',
    y='Principal Component 2',
    hue='completion_status',
    data=pca_df,
    palette='viridis',
    alpha=0.7
)
plt.title('2D PCA Scatter Plot of Principal Components, Colored by Completion Status')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True)
plt.show()

print("2D PCA scatter plot generated.")
```

### Interpreting the 2D PCA Scatter Plot

The scatter plot above visualizes the first two principal components, with data points colored according to their `completion_status`. This plot helps in visually assessing how well the two classes ('Completed' and 'Dropped') are separated in this reduced 2-dimensional space. If distinct clusters are visible for each class, it suggests that these principal components could be effective in distinguishing between the two groups. It also gives an idea of the inherent separability of the data.

### Summary of PCA Observations

Based on the Principal Component Analysis (PCA) performed on the full preprocessed feature set (`X`), the following key observations were made:

*   **Explained Variance**: The cumulative explained variance plot showed that a relatively high number of components (approximately 40-50 out of 56 total features) would be needed to explain nearly 100% of the variance in the dataset. *   **2D Visualization of Class Separability**: The scatter plot visualizing the first two principal components, colored by the `completion_status` (Completed vs. Dropped), revealed significant intermingling between the two classes. ### SHAP (SHapley Additive exPlanations)

SHAP values help to explain the output of any machine learning model. It connects optimal credit allocation with local explanations using game theory. For tree-based models like RandomForestClassifier, SHAP has a highly optimized TreeExplainer.

```python
import shap
import matplotlib.pyplot as plt
import numpy as np

# Assuming rf_model is your trained RandomForestClassifier and X is your preprocessed feature set
explainer = shap.TreeExplainer(rf_model)

# Calculate SHAP values
# For binary classification, shap_values typically returns a list of two arrays [shap_values_class_0, shap_values_class_1]
# or a 3D array (num_samples, num_features, num_classes) depending on the SHAP version and model.
# Let's ensure we are getting the correct shape for class 1 (Completed).

# First, get the raw SHAP values. For TreeExplainer, it's often a list of arrays or a 3D array.
raw_shap_values = explainer.shap_values(X)

# Check the type and shape to determine correct indexing
if isinstance(raw_shap_values, list):
    # If it's a list, we take the element corresponding to the 'Completed' class (index 1).
    shap_values_class_1 = raw_shap_values[1]
elif isinstance(raw_shap_values, np.ndarray) and raw_shap_values.ndim == 3:
    # If it's a 3D array (samples, features, classes), we take the slice for the 'Completed' class (index 1).
    shap_values_class_1 = raw_shap_values[:, :, 1]
else:
    # Fallback or error handling if unexpected format
    raise ValueError("Unexpected format for shap_values.")

# Visualize the SHAP summary plot
print("Generating SHAP summary plot...")
shap.summary_plot(
    shap_values_class_1,
    X,
    show=False # Prevent immediate display, we will use plt.show()
)
plt.title('SHAP Summary Plot for Random Forest Classifier (Class: Completed)')
plt.tight_layout()
plt.show()
print("SHAP summary plot generated.")
```

### Interpreting the SHAP Summary Plot (Global Explanations)

The SHAP (SHapley Additive exPlanations) summary plot provides a global understanding of feature importance and impact. Each dot on the plot represents a single data point from the dataset for a specific feature.

*   **Feature Importance**: The features are ordered from top to bottom by their global importance, meaning features at the top have a larger average absolute SHAP value across all instances, thus contributing more to the model's predictions.
*   **Impact on Model Output (SHAP Value)**: The position of each dot on the x-axis indicates the SHAP value, which represents the impact of that feature on the model's output for that specific instance. Positive SHAP values (to the right) indicate that the feature pushes the prediction higher (e.g., towards 'Completed'), while negative SHAP values (to the left) indicate it pushes the prediction lower (e.g., towards 'Dropped').
*   **Feature Value (Color)**: The color of the dots typically indicates the value of the feature for that instance (e.g., red for high values, blue for low values). This helps to understand whether a high or low value of a feature contributes to a positive or negative prediction. For example, if red dots are concentrated on the positive side of a feature's SHAP values, it means high values of that feature tend to increase the likelihood of the 'Completed' class.
*   **Spread and Density**: The spread of the dots along the x-axis for a given feature shows the range of its impact. The density of the dots indicates how many instances have similar SHAP values for that feature.

By analyzing this plot, we can identify which features are most important overall and how their different values influence the model's predictions for the target class.

### 4.3 Model Evaluation Function

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

def evaluate_model(model_name, y_true, y_pred, y_prob=None):
    """Evaluates a classification model and prints key metrics."""
    print(f"### Evaluation for {model_name}")

    # Define target names for clearer reports and plots
    # 'Dropped' corresponds to 0, 'Completed' corresponds to 1
    target_names = ['Dropped', 'Completed']

    # The positive label should match the label in y_true and y_pred
    # Since y_true and y_pred are now numerical labels (0 and 1),
    # we use 1 as the pos_label for 'Completed'.
    positive_label = 1 # 'Completed' is encoded as 1

    # Calculate metrics. Ensure pos_label matches the actual label value for the positive class.
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, pos_label=positive_label)
    recall = recall_score(y_true, y_pred, pos_label=positive_label)
    f1 = f1_score(y_true, y_pred, pos_label=positive_label)

    print(f"Accuracy: {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall: {recall:.4f}")
    print(f"F1-Score: {f1:.4f}")

    if y_prob is not None:
        # roc_auc_score requires probabilities of the positive class (1 for 'Completed')
        # y_prob[:, 1] assumes that the second column corresponds to the probability of class 1.
        # This is typically correct if classes are ['Dropped', 'Completed'] or 0/1 numeric.
        roc_auc = roc_auc_score(y_true, y_prob[:, 1])
        print(f"ROC-AUC: {roc_auc:.4f}")

    print("\nClassification Report:")
    print(classification_report(y_true, y_pred, target_names=target_names))

    print("\nConfusion Matrix:")
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
                xticklabels=target_names, yticklabels=target_names)
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.title(f'Confusion Matrix for {model_name}')
    plt.show()

print("Model evaluation function 'evaluate_model' defined.")
```

### 4.4 Model Training and Evaluation

#### 4.4.7 Hyperparameter Tuning for SVM (GridSearchCV)

```python
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import precision_score, make_scorer

# Define the parameter grid for GridSearchCV
# Focusing on C, kernel, and gamma which are key for SVC performance
param_grid = {
    'C': [0.1, 1, 10, 100], # Regularization parameter
    'kernel': ['rbf', 'linear'], # Kernel type
    'gamma': ['scale', 'auto', 0.1, 1] # Kernel coefficient for 'rbf', 'poly' and 'sigmoid'
}

# Define a custom scorer to optimize for precision of the 'Dropped' class (label 0)
# Since 'Dropped' is the minority class and our primary concern for recall, we explicitly target it.
# Our y is mapped: 'Completed': 1, 'Dropped': 0.
precision_scorer_dropped = make_scorer(precision_score, pos_label=0)

# Initialize GridSearchCV
# We use the leakage-free training data (X_train_leakage_free, y_train_leakage_free)
# and aim to optimize for the precision of the 'Dropped' class.
# class_weight='balanced' is retained to handle the overall imbalance.
grid_search_svm = GridSearchCV(
    SVC(random_state=42, class_weight='balanced', probability=True), # probability=True for ROC-AUC
    param_grid,
    scoring=precision_scorer_dropped,
    cv=5, # 5-fold cross-validation
    n_jobs=-1, # Use all available CPU cores
    verbose=2
)

# Fit GridSearchCV to the training data
print("Starting GridSearchCV for SVM...")
grid_search_svm.fit(X_train_leakage_free, y_train_leakage_free)

print("GridSearchCV for SVM completed.")

# Get the best parameters and best score
best_params_svm = grid_search_svm.best_params_
best_score_svm = grid_search_svm.best_score_

print(f"\nBest Hyperparameters for SVM: {best_params_svm}")
print(f"Best Cross-validation Precision (for 'Dropped' class): {best_score_svm:.4f}")

# Get the best estimator
best_svm_model = grid_search_svm.best_estimator_

# Make predictions with the best model
y_pred_best_svm = best_svm_model.predict(X_test_leakage_free)
y_prob_best_svm = best_svm_model.predict_proba(X_test_leakage_free)

# Evaluate the best model
evaluate_model('Tuned SVM Classifier (Leakage-Free)', y_test_leakage_free, y_pred_best_svm, y_prob_best_svm)
```

#### 4.4.8 Adjusting Decision Thresholds

After training a classification model that outputs probabilities, the default decision threshold is typically 0.5. This means if the predicted probability of the positive class is above 0.5, the instance is classified as positive. However, for imbalanced datasets or when certain types of errors (false positives vs. false negatives) have different costs, adjusting this threshold can significantly impact model performance metrics like precision and recall.

In our case, we are particularly interested in improving the precision for the 'Dropped' class (which is our minority class, encoded as 0). By lowering the threshold for classifying an instance as 'Dropped', we might increase its recall (identifying more actual dropouts), but potentially at the cost of precision (more false positives). Conversely, raising the threshold for 'Dropped' would make the model more conservative, leading to higher precision but lower recall.

We will explore different thresholds for the 'Dropped' class (0) to find a better balance for our specific problem, aiming to improve precision while maintaining reasonable recall.

```python
from sklearn.metrics import precision_recall_curve, auc
import matplotlib.pyplot as plt

# Get prediction probabilities for the 'Dropped' class (label 0)
# y_prob_best_svm[:, 0] gives probabilities for the 'Dropped' class
y_probs_dropped = best_svm_model.predict_proba(X_test_leakage_free)[:, 0]

# Calculate precision-recall pairs for various probability thresholds
precision_all, recall_all, thresholds = precision_recall_curve(y_test_leakage_free, y_probs_dropped, pos_label=0)

# Plot the Precision-Recall curve
plt.figure(figsize=(10, 7))
plt.plot(recall_all, precision_all, label=f'Precision-Recall curve (AUC = {auc(recall_all, precision_all):.2f})')
plt.xlabel('Recall (for Dropped class)')
plt.ylabel('Precision (for Dropped class)')
plt.title('Precision-Recall Curve for Tuned SVM Classifier (Dropped Class)')
plt.grid(True)
plt.legend()
plt.show()

print("Precision-Recall curve for the 'Dropped' class generated.")
```

The Precision-Recall curve illustrates the trade-off between precision and recall for different decision thresholds. A high AUC (Area Under the Curve) suggests a good performance across all thresholds. We can use this curve to visually identify a threshold that provides a desirable balance for our business objectives. Since our primary goal is to improve precision for the 'Dropped' class, we might look for a point on the curve where precision is higher, even if it means a slight decrease in recall.
